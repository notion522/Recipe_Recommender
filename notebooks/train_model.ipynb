```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
import joblib
import os
import ast

```

```python
df = pd.read_csv("../data/1_Recipe_csv.csv")
df.head()
```

```python
df['ingredients'] = df['ingredients'].apply(ast.literal_eval)
df['ingredients_text'] = df['ingredients'].apply(lambda x: ' '.join(x))

```

```python
df.head()
```

```python
X=df['ingredients_text']
y=df['recipe_title']
```

```python
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000, min_df=5)
X_vec = vectorizer.fit_transform(X)
```

```python
# X_vec.shape
# X_vec
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_vec, y, test_size=0.2, random_state=42)
```

```python

from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train, y_train)
```

```python

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
os.makedirs("../models", exist_ok=True)

joblib.dump(model, "../models/recipe_model.pkl")
joblib.dump(vectorizer, "../models/vectorizer.pkl")

print("✅ Model and vectorizer saved successfully!")
```
